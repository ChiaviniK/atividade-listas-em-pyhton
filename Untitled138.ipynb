```python
# parte 1
produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
precos = [49.90, 129.90, 219.90]

for i,j in zip(produtos, precos):
  print(f"o produto {i} custa R$ {j}")
```

```python
# parte 2
novo_produto = input("Informe o nome do novo produto")
novo_preco = float(input("informe o novo valor"))

produtos.append(novo_produto)
precos.append(novo_preco)

print(f"o produto {novo_produto} foi cadastrado com sucesso")
for i,j in zip(produtos, precos):
  print(f"o produto {i} custa R$ {j}")
```

```python
# parte 3
produtos.pop()
precos.pop()

for i,j in zip(produtos, precos):
  print(f"o produto {i} custa R$ {j}")
```
